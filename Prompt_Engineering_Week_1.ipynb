# Experiment 1: Generating API Keys in OpenAI and Google GenAI

### Aim
To create API keys for OpenAI and Google GenAI (Gemini) platforms and understand how they are used to access AI models through applications.

```python
from google.colab import userdata
import google.generativeai as genai

GOOGLE_API_KEY = userdata.get('geminikey')
genai.configure(api_key=GOOGLE_API_KEY)
model = genai.GenerativeModel('gemini-2.5-flash')

print('Google GenAI API configured successfully.')
```

# Experiment 2: Baseline Prompt and Enhanced Prompt

### Aim
To understand the difference between a Baseline Prompt and an Enhanced Prompt by comparing the quality of AI-generated responses.

```python
print('BASELINE PROMPT')

baseline_prompt = '''
Write a one-paragraph biography of Ada Lovelace.
'''

response = model.generate_content(baseline_prompt)
print(response.text)
```

```python
print('ENHANCED PROMPT')

enhanced_prompt = '''
You are a Computer Science professor.

Write a biography of Ada Lovelace.

Include:
- Early life
- Major contributions
- Why she is called the first programmer
- Legacy

Use markdown headings.
Limit the answer to 200 words.
Use simple English.
'''

response = model.generate_content(enhanced_prompt)
print(response.text)
```

```python
print('PROMPT COMPARISON')
print('-' * 40)
print('Baseline Prompt: Short instruction with little context.')
print('Enhanced Prompt: Detailed instruction with role, format and constraints.')
```

### Result

The enhanced prompt produced a more structured, complete and student-friendly response.

### Conclusion

Enhanced prompts provide context, role, constraints and formatting instructions, which generally produce better-quality AI responses.

# Experiment 3: Iterative Refinement of Prompts

### Aim
To understand the importance of iterative prompt refinement by improving a text summarization task through multiple rounds of prompt engineering.

```python
prompts = [
    'Summarize the plot of Shakespeare\'s play Romeo and Juliet in two sentences.',
    '''Summarize the plot of Shakespeare's play Romeo and Juliet in exactly two concise and formal sentences suitable for high school literature students.''',
    '''Summarize the plot of Shakespeare's play Romeo and Juliet in exactly two formal sentences. Include the setting (Verona, Italy) and emphasize the themes of love, fate, and family conflict.'''
]

for i, prompt in enumerate(prompts, start=1):
    print('=' * 60)
    print(f'ITERATION {i}')
    print('=' * 60)
    response = model.generate_content(prompt)
    print(response.text)
    print()
```

```python
print('ITERATION COMPARISON')
print('-' * 40)
print('Iteration 1: General summary')
print('Iteration 2: Added length and formal style')
print('Iteration 3: Added setting and important themes')
```

### Result

Each refinement improved the clarity, completeness and relevance of the generated summary.

### Conclusion

Improving a prompt step by step with specific instructions produces more structured and informative AI responses.

# Experiment 4: Diagnosing and Refining Prompt Failures

### Aim
To identify common prompt failures caused by ambiguity, missing context, contradictory instructions, or formatting issues, and improve the prompts to obtain accurate AI-generated responses.

```python
prompts = [
    {
        'title': '1. Vague Prompt',
        'prompt': 'Tell me about Python.'
    },
    {
        'title': '2. Contradictory Prompt',
        'prompt': 'Explain Python in one sentence and provide a detailed explanation with examples.'
    },
    {
        'title': '3. Refined Prompt',
        'prompt': '''
Explain Python programming for beginners.

Instructions:
- Write exactly five bullet points.
- Keep each point under 20 words.
- Mention:
  * What Python is
  * Major features
  * Common applications
  * Advantages
  * One real-world example
'''
    }
]

for item in prompts:
    print('=' * 70)
    print(item['title'])
    print('=' * 70)
    response = model.generate_content(item['prompt'])
    print(response.text)
    print()
```

```python
print('PROMPT FAILURE ANALYSIS')
print('-' * 50)
print('Vague Prompt         -> Missing specific requirements')
print('Contradictory Prompt -> Conflicting requirements')
print('Refined Prompt       -> Clear instructions and constraints')
```

### Result

The refined prompt produced a clearer and better-structured response than the vague and contradictory prompts.

### Conclusion

Prompt failures can be reduced by providing clear instructions, sufficient context, examples and output-format requirements.
