## Experiment 11: Building a Simple LCEL Chain Using LangChain Expression Language (LCEL)

**Aim:** To build a simple LCEL chain.


````python
!pip -q install langchain langchain-google-genai
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from google.colab import userdata
key=userdata.get("geminikey")
llm=ChatGoogleGenerativeAI(model="gemini-2.5-flash",google_api_key=key)
prompt=ChatPromptTemplate.from_template("Summarize the following text in 50 words.\n{text}")
chain=prompt | llm | StrOutputParser()
text="Artificial Intelligence enables computers to perform tasks requiring human intelligence such as learning, reasoning and decision making."
print(chain.invoke({"text":text}))
````

## Experiment 12: Basic Data Indexing for Retrieval-Augmented Generation (RAG)

**Aim:** To perform basic RAG data indexing.


````python
!pip -q install langchain langchain-community langchain-google-genai faiss-cpu
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_core.documents import Document
from langchain_community.vectorstores import FAISS
from google.colab import userdata
key=userdata.get("geminikey")
docs=[Document(page_content="Artificial Intelligence enables machines to perform intelligent tasks."),Document(page_content="Machine Learning is a subset of Artificial Intelligence."),Document(page_content="Deep Learning uses neural networks for solving complex problems.")]
chunks=RecursiveCharacterTextSplitter(chunk_size=200,chunk_overlap=20).split_documents(docs)
print("Number of Chunks:",len(chunks))
emb=GoogleGenerativeAIEmbeddings(model="models/embedding-001",google_api_key=key)
db=FAISS.from_documents(chunks,emb); print("Vector Store Created Successfully")
for d in db.similarity_search("What is Machine Learning?",k=2): print(d.page_content)
````

## Experiment 13: Constructing and Running a Basic Retrieval-Augmented Generation (RAG) Chain

**Aim:** To construct and run a basic RAG chain.


````python
!pip -q install langchain langchain-community langchain-google-genai faiss-cpu
from langchain_google_genai import ChatGoogleGenerativeAI,GoogleGenerativeAIEmbeddings
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_core.documents import Document
from langchain_community.vectorstores import FAISS
from google.colab import userdata
key=userdata.get("geminikey")
docs=[Document(page_content="Artificial Intelligence is the science of building intelligent systems. Machine Learning is a subset of AI. Deep Learning is a subset of Machine Learning."),Document(page_content="NLP enables computers to understand and generate human language."),Document(page_content="Computer Vision enables machines to recognize images, objects and videos.")]
chunks=RecursiveCharacterTextSplitter(chunk_size=200,chunk_overlap=20).split_documents(docs)
emb=GoogleGenerativeAIEmbeddings(model="models/embedding-001",google_api_key=key)
store=FAISS.from_documents(chunks,emb); retriever=store.as_retriever(search_kwargs={"k":2})
llm=ChatGoogleGenerativeAI(model="gemini-2.5-flash",temperature=0,google_api_key=key)
q="What is Machine Learning?"; context="\n".join(d.page_content for d in retriever.invoke(q))
print("RAG Answer:"); print(llm.invoke("Answer only using this context:\n"+context+"\nQuestion: "+q).content)
print("\nWithout Retrieval:"); print(llm.invoke(q).content)
````

## Experiment 14: Building a Simple LLM Agent Using LangChain and Tool Calling

**Aim:** To build a simple calculator tool for an LLM-agent workflow.


````python
!pip -q install langchain langchain-google-genai langchain-community
from langchain.tools import Tool
def calculator(expression):
 allowed=set("0123456789+-*/(). "); assert set(expression)<=allowed
 return str(eval(expression,{"__builtins__":{}},{}))
tool=Tool(name="Calculator",func=calculator,description="Performs basic mathematical calculations.")
print("Calculator Tool Created Successfully"); print("Example calculation:",tool.invoke("25-16"))
````
